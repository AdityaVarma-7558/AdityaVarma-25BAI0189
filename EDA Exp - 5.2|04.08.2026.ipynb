```python
import numpy as np
import pandas as pd
import seaborn as sns
from matplotlib import pyplot as plt
from collections import Counter
from sklearn.preprocessing import MinMaxScaler, StandardScaler, RobustScaler
import scipy.stats as stats
data = {
'Age': [25, 30, 45, 35, 22],
'Salary': [50000, 64000, 120000, 85000, 1200000] # 1,200,000 is an outlier
}
df = pd.DataFrame(data)
print("Original Data:\n", df, "\n")
```

```python
min_max_scaler = MinMaxScaler()
df_min_max = pd.DataFrame(
min_max_scaler.fit_transform(df),
columns=df.columns
)
print("Min-Max Scaled Data:\n", df_min_max, "\n")
```

```python
standard_scaler = StandardScaler()
df_standard = pd.DataFrame(
standard_scaler.fit_transform(df),
columns=df.columns
)
print("Standardized Data:\n", df_standard, "\n")
```

```python
data = {
    'StoreSales': [1200, 1500, 1100, 1350, 95000, 1400, 1250, 1600, 1300, 45000]
}
df = pd.DataFrame(data)
Q1 = df['StoreSales'].quantile(0.25)
Q3 = df['StoreSales'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
outliers_iqr = df[(df['StoreSales'] < lower_bound) | (df['StoreSales'] > upper_bound)]
print(outliers_iqr)
```

```python
from scipy.stats import median_abs_deviation
data = {
    'HousePrices': [250000, 270000, 310000, 290000, 260000, 8500000, 280000]
}
df = pd.DataFrame(data)

median = df['HousePrices'].median()
mad = median_abs_deviation(df['HousePrices'])

mod_z_scores = 0.6745 * (df['HousePrices'] - median) / mad
outliers_mod_z = df[abs(mod_z_scores) > 3.5]
print(outliers_mod_z)
```
