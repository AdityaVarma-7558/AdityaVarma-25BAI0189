```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
df = pd.read_csv("/content/WineQT.csv")
print("First 3 rows of the dataset:")
display(df.head(3))
print("\nLast 3 rows of the dataset:")
display(df.tail(3))
```

```python
print("Data types of each column:")
display(df.dtypes)
print("\nDescriptive statistics of the dataset:")
display(df.describe())
```

```python
print("Missing values before handling (NaN):")
display(df.isnull().sum())
for col in df.select_dtypes(include=np.number).columns:
    if df[col].isnull().any():
        df[col] = df[col].fillna(df[col].mean())
print("\nMissing values after handling (NaN):")
display(df.isnull().sum())
```

```python
mean_alcohol = df["alcohol"].mean()
median_alcohol = df["alcohol"].median()
mode_alcohol = df["alcohol"].mode()[0]
print(f"Alcohol - Mean: {mean_alcohol:.2f}, Median: {median_alcohol:.2f}, Mode: {mode_alcohol:.2f}")
```

```python
plt.figure(figsize=(10, 6))
df['quality'].value_counts().sort_index().plot(kind='bar')
plt.title("Number of Wines by Quality")
plt.ylabel("Number of Wines")
plt.xlabel("Quality")
plt.xticks(rotation=0)
plt.show()
```

```python
sns.FacetGrid(df, height=5, aspect=1.5).map(sns.histplot, "fixed acidity", kde=True).add_legend()
plt.title("Distribution of Fixed Acidity")
plt.xlabel("Fixed Acidity")
plt.ylabel("Density")
plt.show()
```

```python
sns.FacetGrid(df, height=5, aspect=1.5).map(sns.histplot, "alcohol", kde=True).add_legend()
plt.title("Distribution of Alcohol Content")
plt.xlabel("Alcohol Content")
plt.ylabel("Density")
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.boxplot(x="quality", data=df)
plt.title("Box Plot for Wine Quality")
plt.xlabel("Quality")
plt.ylabel("Value")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(df["alcohol"], df["quality"])
plt.title("Scatter Plot for Alcohol Content vs Quality")
plt.xlabel("Alcohol Content")
plt.ylabel("Quality")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x="quality", y="volatile acidity", data=df)
plt.title("Box Plot for Volatile Acidity by Wine Quality")
plt.xlabel("Quality")
plt.ylabel("Volatile Acidity")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x="quality", y="sulphates", data=df)
plt.title("Box Plot for Sulphates by Wine Quality")
plt.xlabel("Quality")
plt.ylabel("Sulphates")
plt.show()
```

```python
sns.pairplot(df, vars=["fixed acidity", "alcohol", "quality"], kind="reg")
plt.suptitle("Pair Plot: Fixed Acidity, Alcohol, and Quality (Regression)", y=1.02)
plt.show()
```

```python
sns.set(style="ticks", color_codes=True)
sns.pairplot(
    df,
    height=2,
    vars=["fixed acidity", "pH", "alcohol"],
    hue="quality",
    palette="viridis"
)
plt.suptitle("Pair Plot: Fixed Acidity, pH, and Alcohol (Hue by Quality)", y=1.02)
plt.show()
```

```python
from scipy import stats
corr_alcohol_quality = stats.pearsonr(df["alcohol"], df["quality"])
print(f"Pearson correlation between Alcohol and Quality:\n  p-value:\t {corr_alcohol_quality[1]:.3f}\n  correlation:\t {corr_alcohol_quality[0]:.3f}")
```

```python
correlation = df.corr(method="pearson", numeric_only=True)
print("\nCorrelation Matrix:")
display(correlation)
plt.figure(figsize=(12, 10))
sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt=".2f",
    linewidths=.5,
    cbar_kws={'label': 'Correlation Coefficient'}
)
plt.title("Correlation Matrix of Wine Dataset Features")
plt.show()
```
